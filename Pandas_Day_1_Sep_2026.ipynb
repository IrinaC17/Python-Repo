- df.iloc[row index(es), column index(es)]

```python
# mark of the student at index 25
df.iloc[25, 3]
```

```python
# name and gender of the student at index 3
df.loc[3, ["name", "gender"]]
```

```python
df.iloc[3, [1,4]]
```

# How to filter the rows.

df[condition]

```python
df[df["mark"]>90]
```

```python
df[df["mark"]==90]
```

```python
df[df["class"]=="Seven"]
```

```python
df[df["gender"]=="female"]
```

- df[ (condition1) & (condition2)]

```python
# return male students at class four
df[(df["gender"]=="male") & (df["class"]=="Four")]
```

```python
# find the info for the student who has less than 90 mark  and female
```

```python
df[(df["mark"]<90) & (df["gender"]=="female")]
```

```python
# average mark of the female student
df[df["gender"]=="female"]["mark"].mean()
```

```python
# mark > 90 or male student
df[(df["mark"]>90) | (df["gender"]=="male")]
```

- df.loc[condition, column(s)]

```python
df.loc[df["mark"]>90, :]
# : means all columns
```

```python
df.iloc[df["mark"]>90, :]
```

- df.loc[(conditon1) & (conditon2), column(s)]

```python
df.loc[(df["mark"]>80) & (df["class"]=="Five"), ["name", "mark"]]
# : means all columns
```

```python
df.filter(items=['name', 'mark'])
```

```python
df.filter(like='ma').head()
```

```python
df.query("mark > 80")
```

```python
df.query('gender == "female"')
```
